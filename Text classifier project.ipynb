```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.pipeline import make_pipeline

cats = ["sci.med", "sci.space", "rec.sport.hockey", "comp.graphics"]
opts = dict(categories=cats, remove=("headers", "footers", "quotes"))
train = fetch_20newsgroups(subset="train", **opts)
test = fetch_20newsgroups(subset="test", **opts)

model = make_pipeline(TfidfVectorizer(stop_words="english"),
                      LogisticRegression(max_iter=1000))
model.fit(train.data, train.target)
print(classification_report(test.target, model.predict(test.data),
                            target_names=test.target_names))
```

```python
from sklearn.naive_bayes import MultinomialNB

model_nb = make_pipeline(TfidfVectorizer(stop_words="english"),
                          MultinomialNB())
model_nb.fit(train.data, train.target)
print(classification_report(test.target, model_nb.predict(test.data),
                            target_names=test.target_names))
```

```python
my_texts = [
    "The doctor prescribed new medication for the patient",
    "The team scored in the final period of the game",
]
predictions = model.predict(my_texts)
for text, pred in zip(my_texts, predictions):
    print(text, "->", train.target_names[pred])
```
